```python
import requests
from bs4 import BeautifulSoup
import json
import pandas
```

```python
url = "https://www.espncricinfo.com/records/tournament/team-match-results/icc-men-s-t20-world-cup-2022-23-14450"

response = requests.get(url=url)

cric_data_response = response.text

soup = BeautifulSoup(cric_data_response, "html.parser")

match_results = soup.find_all(name= 'a', class_ = 'ds-inline-flex ds-items-start ds-leading-none')

item_list = []
for item in match_results:
    if "full-scorecard" in item.get('href'):
        item_list.append(f"https://www.espncricinfo.com{item.get('href')}")
```

```python

tables = pandas.read_html(item_list[0])

df_score = pandas.DataFrame(tables[0])

df_score = df_score.iloc[:-2]



```
